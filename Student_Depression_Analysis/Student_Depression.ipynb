# **Exploratory Analysis**

**Depressed people percentage by Gender**

```python
depressed_people = df[df['Depression'] == 1]
depressed_people_gender = (depressed_people.groupby('Gender')['Depression'].count() / depressed_people.shape[0]) * 100
print(round(depressed_people_gender, 1))
```

**Family History of depressed people**

```python
familyH_suicidalT_depressed = (depressed_people.groupby('Family History of Mental Illness')['Depression'].count() / depressed_people.shape[0])*100
print(round(familyH_suicidalT_depressed),1)
```

**Depressed people by Age**

```python
age_depression_count = depressed_people.groupby('Age')['Depression'].sum()

print(age_depression_count)
```

**Dietary Habits of Depressed People and non-Depressed People**

```python
depressed_people_dietary_habits = ((depressed_people.groupby('Dietary Habits')['Depression'].count())/depressed_people.shape[0])*100
print(round(depressed_people_dietary_habits),1)
```

```python
non_depressed_people = df[df['Depression'] == 0]

non_depressed_people_dietary_habits = ((non_depressed_people.groupby('Dietary Habits')['Depression'].count())/non_depressed_people.shape[0])*100
print(round(non_depressed_people_dietary_habits),1)
```

**Is There any relation between CGPA and Sleep Duration?**

```python
gcpa_sleep_duration = df.groupby('Sleep Duration')['CGPA'].mean()
print(gcpa_sleep_duration.round(2))
```

**Is there any relation between Depression and Sleep Duration?**

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Sleep Duration', hue='Depression', data=df, palette='Set1')
plt.title('Sleep Duration vs Depression')
plt.xlabel('Sleep Duration')
plt.ylabel('Count')
plt.legend(title='Depression')
plt.show()
```

**Correlation Analysis**

```python
numerical_data = df.select_dtypes(include=['float64', 'int64'])  # Select only numerical columns
correlation_matrix = numerical_data.corr()  # Calculate the correlation matrix

# Plot the heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True)
plt.title('Correlation Matrix')
plt.show()
```
