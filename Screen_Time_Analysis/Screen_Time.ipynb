# **Introduction**

This Analysis was made using the Screen Time vs Mental Wellness Survey - 202 dataset, available on the website kaggle.com

The motivation is to analyze the variables and the relation between the screen time and mental wellness of the people, getting important insights

This can be important to improve the mental wellness and prevent diseases.

# **Importing libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# **Load Dataset**

```python
df = pd.read_csv('ScreenTime_vs_MentalWellness.csv')
```

```python
sns.histplot(df["screen_time_hours"], bins=20, kde=True)
plt.title("Distribuição de Horas de Tela")
plt.xlabel("Horas de tela por dia")
plt.ylabel("Participantes")
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
sns.histplot(df["sleep_quality_1_5"], bins=np.arange(0.5, 5.5 + 1, 1), ax=axes[0], kde=True)
axes[0].set_title("Distribuição do Sono")

sns.histplot(df["stress_level_0_10"], bins=10, ax=axes[1], kde=True)
axes[1].set_title("Distribuição do Stress")

sns.histplot(df["productivity_0_100"], bins=10, ax=axes[2], kde=True)
axes[2].set_title("Distribuição da Produtividade")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(y="occupation", data=df, color='skyblue', order=df["occupation"].value_counts().index)
plt.title("Distribuição por Ocupação")
plt.show()
```

```python
bins = [0, 18, 25, 35, 50, 100]
labels = ["<18", "18-25", "26-35", "36-50", "50+"]
df["faixa_etaria"] = pd.cut(df["age"], bins=bins, labels=labels)

sns.countplot(x="faixa_etaria", color='cyan', data=df, order=labels)
plt.title("Distribuição Etária")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="screen_time_hours",
    y="stress_level_0_10",

)
plt.title("Screen time vs Stress")
plt.show()
```
